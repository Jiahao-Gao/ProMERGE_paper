### Permute

```python
for f in [
    "BioPlex-HCT116.tsv",
    "BioPlex-HEK293T.tsv",
    "SECMX-HCT116.tsv",
    "SECMX-HEK293T.tsv"
]:
    df = pd.read_csv(
        f"../../result/HEK293T_HCT116/input/{f}",
        sep="\t", index_col=0
    )
    df.index = np.random.permutation(df.index)
    df = df.sort_index()
    df.to_csv(
        f"../../result/HEK293T_HCT116_permute/input/{f}",
        sep="\t"
    )
```

### Stats for future filtering

```python
# load genes in each experiment
ap_hct = set(pd.read_csv("../../result/HEK293T_HCT116_permute/input/BioPlex-HCT116.tsv", sep="\t", index_col=0).index)
ap_hek = set(pd.read_csv("../../result/HEK293T_HCT116_permute/input/BioPlex-HEK293T.tsv", sep="\t", index_col=0).index)
sec_hct = set(pd.read_csv("../../result/HEK293T_HCT116_permute/input/SECMX-HCT116.tsv", sep="\t", index_col=0).index)
sec_hek = set(pd.read_csv("../../result/HEK293T_HCT116_permute/input/SECMX-HEK293T.tsv", sep="\t", index_col=0).index)
```

```python
# total possible pairs in each cell line
hct_genes = ap_hct | sec_hct
hct_pairs = len(hct_genes) * (len(hct_genes)-1) // 2

hek_genes = ap_hek | sec_hek
hek_pairs = len(hek_genes) * (len(hek_genes)-1) // 2
```

```python
# top 1%, add other when needed
stat = pd.DataFrame(
    [
        [int(hct_pairs*0.01), int(hek_pairs*0.01)]
    ],
    index=["Percent1"],
    columns=["HCT116", "HEK293T"]
)
stat.to_csv("../../result/HEK293T_HCT116_permute/input/pair_stat.tsv", sep="\t")
```
