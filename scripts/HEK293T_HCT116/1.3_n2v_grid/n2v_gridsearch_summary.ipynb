```python
import argparse
import glob
import pandas as pd
```

```python
rows = pd.concat([
    pd.read_csv(file, sep="\t")
    for file in sorted(
        glob.glob("../../../result/HEK293T_HCT116/n2v_gridsearch/*.tsv")
    )
])
```

```python
summary = rows.groupby(
    ["config_id", "p", "q", "dimensions", "walk_length", "num_walks"],
    as_index=False
)["auroc"].mean()
summary = summary.sort_values("auroc", ascending=False)
summary.to_csv(
    "../../../result/HEK293T_HCT116/n2v_gridsearch_summary.tsv",
    sep="\t", index=False
)
```
