### Stats for future filtering

```python
ap_hct = set(pd.read_csv("../../result/HEK293T_HCT116_mod3/input/BioPlex-HCT116.tsv", sep="\t", index_col=0).index)
ap_hek = set(pd.read_csv("../../result/HEK293T_HCT116_mod3/input/BioPlex-HEK293T.tsv", sep="\t", index_col=0).index)
sec_hct = set(pd.read_csv("../../result/HEK293T_HCT116_mod3/input/SECMX-HCT116.tsv", sep="\t", index_col=0).index)
sec_hek = set(pd.read_csv("../../result/HEK293T_HCT116_mod3/input/SECMX-HEK293T.tsv", sep="\t", index_col=0).index)
tpca_hct = set(pd.read_csv("../../result/HEK293T_HCT116_mod3/input/TPCA-HCT116.tsv", sep="\t", index_col=0).index)
tpca_hek = set(pd.read_csv("../../result/HEK293T_HCT116_mod3/input/TPCA-HEK293T.tsv", sep="\t", index_col=0).index)
```

```python
# total possible pairs in each cell line
hct_genes = ap_hct | sec_hct | tpca_hct
hct_pairs = len(hct_genes) * (len(hct_genes)-1) // 2

hek_genes = ap_hek | sec_hek | tpca_hek
hek_pairs = len(hek_genes) * (len(hek_genes)-1) // 2
```

```python
# top 1%, add other when needed
stat = pd.DataFrame(
    [
        [int(hct_pairs*0.01), int(hek_pairs*0.01)]
    ],
    index=["Percent1"],
    columns=["HCT116", "HEK293T"]
)
stat.to_csv("../../result/HEK293T_HCT116_mod3/input/pair_stat.tsv", sep="\t")
```
