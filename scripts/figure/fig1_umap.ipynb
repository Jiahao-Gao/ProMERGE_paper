```python
import umap
import pandas as pd
import matplotlib.pyplot as plt
```

```python
emb1 = pd.read_csv("../../result/HEK293T_HCT116/coemb/promergeclub_HCT116anchor/HCT116/HCT116_latent.tsv", sep="\t", index_col=0)
emb2 = pd.read_csv("../../result/HEK293T_HCT116/coemb/promergeclub_HCT116anchor/HEK293T/HEK293T_latent.tsv", sep="\t", index_col=0)
genes = sorted(set.intersection(set(emb1.index), set(emb2.index)))
emb1 = emb1.loc[genes]
emb2 = emb2.loc[genes]
```

### UMAP base

```python
reducer = umap.UMAP(random_state=1)
embedding = reducer.fit_transform(emb1)
```

```python
plt.style.use("/home/jig049/misc/style.mplstyle")
fig, ax = plt.subplots(figsize=(1, 1), dpi=300)
plt.scatter(
    embedding[:, 0],
    embedding[:, 1],
    s=1,
    marker='.',
    color='#9fc6dd',
    alpha=0.1
)
plt.xticks([])
plt.yticks([])
ax.spines[['right', 'top', 'bottom', 'left']].set_visible(False)

plt.savefig("./fig1_umap_base.pdf", bbox_inches=None, transparent=True)
```

### UMAP query

```python
reducer = umap.UMAP(random_state=1)
embedding = reducer.fit_transform(emb2)
```

```python
plt.style.use("/home/jig049/misc/style.mplstyle")
fig, ax = plt.subplots(figsize=(1, 1), dpi=300)
plt.scatter(
    embedding[:, 0],
    embedding[:, 1],
    s=1,
    marker='.',
    color='#dba1a1',
    alpha=0.1
)
plt.xticks([])
plt.yticks([])
ax.spines[['right', 'top', 'bottom', 'left']].set_visible(False)

plt.savefig("./fig1_umap_query.pdf", bbox_inches=None, transparent=True)
```
